## Sample Arrays

```python
#creating 5 variables that contain random sample arrays
r_1 = np.random.randint(1,21,size=2) #example 1
r_2 = np.random.randint(1,21,size=(1,2)) #example 2
r_3 = np.random.rand(2,1) #example 3
r_4 = np.random.randint(20,101,size=(4,5)) #example 4
r_5 = np.random.rand(4,5) #example 5

#displaying outputs
print(f"r_1: \n",r_1)
print(f"\nr_2: \n",r_2)
print(f"\nr_3: \n",r_3)
print(f"\nr_4: \n",r_4)
print(f"\nr_5: \n",r_5)
```

## Permutations

```python
#printing current value of r_4
print(f"Original r_4: \n",r_4)
```

```python
#shuffling r_4
np.random.shuffle(r_4)
print(f"Shuffled r_4: \n",r_4) #displaying result
```

r_4 was shuffled by moving around the rows within the array, but the the elements stayed in their place in each row.

```python
#creating new variable for new_shuffle
new_shuffle = np.random.permutation(r_4) #permutation to shuffle

#printing new results
print(f"New Shuffle: \n",new_shuffle)
print(f"\nComparison to Original: \n",r_4)
```

### Shuffle vs Permutation
Shuffle changed the array where the rows are moved around but the values stays in place in their rows.
Permutation changed the array by making a new copy of the array, again, moving around the rows but kept the values in place in their rows.
Both remained 2D, the functions worked in a way where it did shuffle, but the confusion that was encountered is the values not shuffling also.

I would use shuffle to modify the ORIGINAL array, and use permutation to modify a copy, leaving the original unchanged.

## Grocery Problem

```python
#creating two new variables
bag1 = np.array([2,3,4,0]) #2 veggies, 3 fruits, 4 snacks, 0 drinks
bag2 = np.array([4,0,1,2]) #4 veggies, 0 fruits, 1 snacks, 2 drinks

#displaying bag1 & bag2
print(f"bag1:",bag1)
print(f"bag2:",bag2)
```

```python
# total items in each bag
bag1_total = np.sum(bag1)
bag2_total = np.sum(bag2)

#displaying output
print(f"The first bag contains {bag1_total} items, and the second bag contains {bag2_total} items.")
```

```python
#total items in each category
category_totals = bag1 + bag2

veg, fruit, snacks, bev = category_totals #categories

#displaying output
print(f"There are {veg} vegetables, {fruit} fruits, {snacks} boxes of snacks, {bev} beverages in the two grocery bags.")
```

## Teamwork

```python
# Ali's report
ali = np.array([5,4,6,4,2]) #doors,Mon,Tue,Wed,Thu,Fri
print("Ali:",ali) #displaying results
```

```python
# Ben's report
ben = np.array([0,5,5,3,4]) #doors,Mon,Tue,Wed,Thu,Fri
print("Ben:",ben) #displaying results
```

```python
# Carla's report
carla = np.array([3,6,0,3,4]) #doors,Mon,Tue,Wed,Thu,Fri
print("Carla:",carla) #displaying results
```

```python
#compiling all three reports into a single array
all_knocks = np.array([ali,ben,carla])
print(f"All Knocks: \n",all_knocks) #displaying results
```

```python
#total number of doors each person knocked
ali_total = np.sum(ali)
ben_total = np.sum(ben)
carla_total = np.sum(carla)

#total as a group
group_total = np.sum(all_knocks)

print(f"Ali knocked on {ali_total} doors.")
print(f"Ben knocked on {ben_total} doors.")
print(f"Carla knocked on {carla_total} doors.")
print(f"The group collectively knocked on {group_total} doors.")
```

```python
#daily totals across the three
daily_totals = np.sum(all_knocks,axis=0) #sum on axis 0
print(f"Daily Totals:",daily_totals)

#rolling total
rolling_knocks = np.cumsum(daily_totals) #cumulative sum
print(f"\nRolling Door-Knock Totals:",rolling_knocks)

#estimated votes
estimated_votes = rolling_knocks*0.5 #half of rolling knocks
print(f"\nEstimated Rolling Votes:",estimated_votes)
```
