## Series

```python
#creating 4 new variables
aa = [1,2,3,4] #list
ab = (5,6,7,8) #tuple
ac = {9,10,11,12} #set
ad = {'first':13,'second':14,'third':15,'fourth':16} #dictionary

#printing outputs
print("aa:",aa)
print("ab:",ab)
print("ac:",ac)
print("ad:",ad)
```

```python
#creating 4 more new variables

#assigning series to previous variables
ser1 = pd.Series(aa)
ser2 = pd.Series(ab)
ser3 = pd.Series(ac)
ser4 = pd.Series(ad)

#printing output
print("ser1: \n",ser1)
print("\nser2: \n",ser2)
print("\nser3: \n",ser3) #TypeError: 'set' type is unordered
print("\nser4: \n",ser4)
```

## DataFrames from Series

```python
#creating new variable for DataFrame1 using ser1
df1 = pd.DataFrame(ser1)
df1 #displaying df1
```

```python
#displaying the last 10 records
df2.tail(10)
```

```python
#displaying information summary
df2.info()
```

### Summary Info for df2
This dataset had 262 rows & 5 columns. All data types are objects, meaning it is text-based.
The columns are named title, description, category, language and actor_name.
Every column has 262 non-null values, so there are no missing values in this dataset.
There is 10.4 KB memory stored, so this tells how large or little the dataset is.

```python
#showing column statistics
df2.describe() #using describe to show column statistics
```

### Column Statistics for df2
Because all the columns are text, describe was able to show the count, unique, top, frequent values.
This is useful when trying to see which columns appear most in the dataset.
It is not useful when attempting a deeper analysis since the columns are text and not numeric.

```python
#creating new variable
df3 = pd.read_csv("northwind_example.csv") #reading csv file
df3 #displaying output
```

```python
#showing column statistics
df3.describe() #using describe to show column statistics
```

### Column Statistics for df3
Using describe was helpful because this dataset has numeric columns.
Because of this, we are able to see certain stats like min, max and mean, letting us see how the numbers are spread out.
This is different from df2, because df2 had text columns and using describe only showed unique values. With df3
we are able to gain more insight since the dataset is numeric, which lets us see more statistical information.

## Coding a DataFrame
