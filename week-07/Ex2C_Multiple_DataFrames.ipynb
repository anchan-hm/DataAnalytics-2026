# Exercise 2.C Working with multiple DataFrames

```python
#importing libraries
import numpy as np
import pandas as pd
```

## Basic Aggregations

```python
#creating a DataFrame
data = {'Product':['A','B','C','A','B','C','A','B','C'],
        'Region':['North','South','East','West','North','South','East','West','North'],
        'Units_Sold':[10,15,20,5,10,15,10,5,15],
        'Price':[100,200,150,100,200,150,100,200,150]}

df_sales = pd.DataFrame(data) #creating DataFrame from the dictionary
df_sales #displaying output
```

```python
#adding a new column
df_sales['Revenue'] = df_sales['Units_Sold']*df_sales['Price'] #creating a column as Revenue
df_sales #displaying output
```

```python
#calculating total units sold across all products & regions
total_units = df_sales['Units_Sold'].sum()

#calculating total revenue across all products & regions
total_rev = df_sales['Revenue'].sum()

print(f"Total Units Sold: ",total_units)
print(f"Total Revenue: ",total_rev)
```

```python
#calculating the average units sold
avg_units = df_sales['Units_Sold'].mean()
avg_units_round = round(avg_units,2) #rounding result

#calculating the average revenue
avg_rev = df_sales['Revenue'].mean()
avg_rev_round = round(avg_rev,2)

print(f"Average Units Sold: ",avg_units_round)
print(f"Average Revenue: ",avg_rev_round)
```

## Grouping Data

```python
#grouping products to calculate total (units sold & revenue)
product_total = df_sales.groupby('Product')[['Units_Sold','Revenue']].sum()
product_total #displaying result
```

```python
#grouping region to calculate average total (units sold & revenue)
region_avg = df_sales.groupby('Region')[['Units_Sold','Revenue']].mean()
region_avg #displaying output
```

```python
#grouping products & region to calculate total (units sold & revenue)
region_product_total = df_sales.groupby(['Product','Region'])[['Units_Sold','Revenue']].sum()
region_product_total #displaying output
```

## Sorting and Ranking Data

```python
#sorting units sold in ascending order
df_sales.sort_values(by='Units_Sold',ascending=True) #smallest to largest
```

```python
#sorting revenue in descending order
df_sales.sort_values(by='Revenue',ascending=False) #largest to smallest
```

```python
#adding a column for revenue ranking
df_sales['Revenue_Rank']=df_sales['Revenue'].rank(ascending=False) #highest revenue is ranked 1
```

```python
df_sales #displaying ranking output
```

## Merging DataFrames

```python
product_data = {'Product':['A','B','C'],
                'Category':['Electronics','Furniture','Appliances']}

df_products = pd.DataFrame(product_data) #creating df from dictionary
df_products #displaying output
```

```python
#merging df_sales with df_products on product column
df_merged = pd.merge(df_sales,df_products,on='Product')
df_merged #displaying output
```

```python
#grouping categories to calculate total rev
category_rev = df_merged.groupby('Category')['Revenue'].sum()
category_rev #displaying output
```
