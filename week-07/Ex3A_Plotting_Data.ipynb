# Exercise 3.A Gettung familiar with Matplotlib

```python
#importing libraries
import numpy as np
import matplotlib.pyplot as plt
```

## Creating Basic Plots

```python
#creating new variables
x = [1,2,3,4,5]
y = [2.,3,5,7,11]

#plotting line graph
plt.plot(x,y)
plt.show() #displaying output
```

```python
#plotting line graph with markers
plt.plot(x,y,marker='o') #adding markers

# adding titles and labels
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')

plt.show() #displaying plot
```

## Styling Plots

```python
#plotting line graph with red dashes
plt.plot(x,y,linestyle='--',color='red') #red dashes
plt.show() #displaying output
```

```python
#plotting line graph with square markers
plt.plot(x,y,marker='s',markersize=10,markerfacecolor='green') #green square markers
plt.show() #displaying output
```

```python
#two different additional markers and colors

#example 1
plt.plot(x,y,marker='^',color='blue') #blue triangles
plt.show() #displaying output
```

```python
#example 2
plt.plot(x,y,marker='D',color='purple') #purple diamonds
plt.show() #displaying output
```

## Advanced Plotting Techniques

```python
#creating data for bar chart
products = ['Product A','Product B','Product C','Product D']
sales = [23,17,35,29]

#plotting bar chart
plt.bar(products,sales)
plt.show() #displaying output
```

```python
#customizing bar chart
plt.bar(products,sales,color='skyblue',width=0.5) #setting color and width

#adding title and labels
plt.title('Product Sales')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.show() #displaying output
```

```python
#creating scatter plot
x = np.random.rand(50)
y = np.random.rand(50)

#plotting scatter plot
plt.scatter(x,y)
plt.show() #displaying output
```

```python
#creating random sized markers
sizes = np.random.rand(50)*100

#customizing scatter plot with sizes & colormap
plt.scatter(x,y,s=sizes,c=sizes,cmap='viridis')

#adding title and labels
plt.title("Customized Scatter Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show() #dispalying output
```

## Subplots

```python
#creating data for sine wave
x = np.linspace(0,10,100)
y = np.sin(x)

#creating subplots
plt.subplot(1,2,1) #1 row, 2 columns, first subplot
plt.plot(x,y)
plt.title('Sine Wave')
plt.xlabel('X Values')
plt.ylabel('sin(x)')
```

```python
#second subplot (histogram)
plt.subplot(1,2,2) # 1 row, 2 columns, second subplot
plt.hist(np.random.randn(1000), bins=30, color='skyblue') #1000 randomly generated numbers, 
plt.title("Histogram of Random Numbers")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.tight_layout() #adjust spacing so title and labels do not overlap
plt.show() #displaying output
```
