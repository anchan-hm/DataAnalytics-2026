## Looping Through Arrays

```python
#creating 2D array
arrA = np.array([[11,12,13],[14,15,16]])
```

```python
#printing array, ensuring it is 2D
print(arrA.ndim)
```

```python
print(arrA)
```

```python
#looping through array and multiply each element by 2
for x in arrA:
        print(x*2)
```

```python
for x in arrA:
    for y in x:
        print(y*2)
```

The only difference is that they start to get seperated from the brackets, but they still end up resulting in the same multiplied number.

```python
for x in arrA:
    for y in x*2:
        print(y)
```
