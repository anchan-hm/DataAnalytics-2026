```python
#importing library
import numpy as np
```

## Generating Random Numbers

```python
# generating random integers (1-20)
r_1 = np.random.randint(1,21,size=2) #creating variable with random numbers into 2 elements
print(f"2 Random Elements: \n",r_1) # displaying results
```

```python
# generating random integers (1-20) SIZE: 1 row, 2 columns
r_2 = np.random.randint(1,21,size=(1,2)) #creating new variable with random numbers (size 1x2)
print(f"Random 1 Row, 2 Columns: \n",r_2) #displaying results
```

```python
# generating random floats (0-1) SIZE: 2 rows, 1 column
r_3 = np.random.rand(2,1) #using rand to create new variable with random floats (2x1)
print(f"Random 2 Rows, 1 Column (floats): \n",r_3) #displaying results
```

```python
# generating random integers (20-100) SIZE: 4 rows, 5 columns
r_4 = np.random.randint(20,101,size=(4,5)) #creating new variable with random numbers (size 4x5)
print(f"Random 4 Rows, 5 Columns: \n",r_4) #displaying results
```

```python
# generating random floats (0-1) SIZE: 4 rows, 5 columns
r_5 = np.random.rand(4,5) # creating new variable with random floats (4x5)
print(f"Random 4 Rows, 5 Columns (floats): \n",r_5) # displaying results
```

## Generating Random Choices from an Array

```python
# creating a 1D array for student names
students = np.array(["Alexus","Luis O.","Naomi",
                     "Alondra","Victoria","Blake",
                     "Corey","Lorah","Lesley",
                     "Sonia","Luis A.","Charles"])

print(f"Sec 12: \n", students) #displaying list of names
```

```python
# creating new variable to call on a student randomly
call_on = np.random.choice(students) #randomly select
print(f"Student to Call on: \n", call_on) #displaying output
```

```python
# creating a new variable to choose a random sampling
breakout_groups = np.random.choice(students, size=(3,4)) #3 rows, 4 columns
print(f"Breakout Groups: \n",breakout_groups) #displaying output
# this method displayed students being repeated and also not showing all students
```

```python
#creating new variable to choose random students with probabilities
students_pvalues = [0.30,0.10,0.05, #0.30 (30%) more likely
                    0.05,0.10,0.10,
                    0.05,0.10,0.05,
                    0.05,0.05,0.00] #0.00 (0%) less likely

call_on_weighted = np.random.choice(students, size=5, p=students_pvalues) #creating random choice
print(f"Students Being Called on: \n",call_on_weighted) #displaying results
```

## Generating a Normal Distribution

```python
#creating new variable for random distribution curve
bell_curve = np.random.normal(10,5, size=(4,5)) # peak curve:10,standdev:5, size:4x5
print(f"Bell Curve: \n",bell_curve) #displaying results
```
