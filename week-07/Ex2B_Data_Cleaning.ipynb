# Exercise 2.B Cleaning up data

```python
#importing libraries
import numpy as np
import pandas as pd
```

```python
#creating a new df
df = pd.read_csv("test_data.csv") #using test_data for df
df #displaying output
```

```python
#showing column statistics
df.describe() #using describe to show column statistics
```

## Filtering a DataFrame

```python
#filtering df to show age GREATER THAN 18
df[df['Age']>18] #age greater than 18
```

```python
#filtering df show only Stores GREATER THAN 50 & LESS THAN 100
df[(df['Store']>50) & (df['Store']<100)] #greater than 50, less than 100
```

## Modifying a DataFrame

```python
#creating new variable to clean
df_cleaned = df.drop(df[df['Store']<0].index) #dropping Stores below 0
df_cleaned #displaying output
```

```python
#cleaning more of the data
df_cleaned = df_cleaned.drop(columns=['City']) #dropping City column
df_cleaned #displaying output
```

```python
#replacing missing values
df_cleaned['Age'] = df_cleaned['Age'].fillna(df_cleaned['Age'].mean()) #replacing missing age values
df_cleaned #displaying results
```

```python
#replacing 0 values
df_cleaned['Store'] = df_cleaned['Store'].replace(0,df_cleaned['Store'].mean()) #replacing 0 values in Store
df_cleaned #displaying output
```

```python
#dropping NaN values
df_cleaned = df_cleaned.dropna() #using dropna to drop NaN
df_cleaned #displaying output
```

```python
#taking ID values as index labels
df_cleaned = df_cleaned.set_index('ID')
df_cleaned = df_cleaned.drop(columns=['ID'],errors='ignore') #ignoring errors and dropping column
df_cleaned
```
