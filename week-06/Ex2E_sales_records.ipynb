This notebook analyzes s small dataset using Python functions.
Each function takes the full dataset as its first argument.

```python
# creating sales_data variable with product, region, and sales amount as tuples
sales_data = [('Laptop', 'East', 1200.00),
              ('Phone', 'West', 850.00),
              ('Tablet', 'East', 400.00),
              ('Monitor', 'South', 650.00),
              ('Keyboard', 'West', 75.00),
              ('Laptop', 'South', 1150.00),
              ('Phone', 'East', 900.00),
              ('Tablet', 'West', 425.00),
              ('Monitor', 'East', 680.00),
              ('Keyboard', 'South', 80.00),
              ('Laptop', 'West', 1300.00),
              ('Phone', 'West', 820.00),]
```

## Lab 1. Total sales by region

```python
# creating a function for total sales in each region
def region_total(data, region): # grabbing total sales for each region
    return sum(amt for _, r, amt in data if r == region) # unpacking
```

```python
# testing for each region
print(region_total(sales_data, 'East')) # testing East region
print(region_total(sales_data, 'West')) # testing West region
print(region_total(sales_data, 'South')) # testing South region
```

## Lab 2. Count of sales by region

```python
# creating a function to count sales for each region
def region_count(data, region): # counting each sale for each region
    return sum(1 for _, r, _ in data if r == region) # unpacking
```

```python
# testing for each region
print(region_count(sales_data, 'East')) # testing East region
print(region_count(sales_data, 'West')) # testing West region
print(region_count(sales_data, 'South')) # testing South region
```

## Lab 3. Average sale amount by region

```python
# creating a function for average sale for each region
def region_average(data, region): # average sale for each region
    total = region_total(data, region) # grabbing total sale per region
    count = region_count(data, region) # grabbing count of each sale per region

    if count == 0: # no record of sale (avoid division by zero error)
        return 0

    return total/count # returning the average sale
```

```python
# testing for each region
print(region_average(sales_data, 'East')) # testing East region
print(region_average(sales_data, 'West')) # testing West region
print(region_average(sales_data, 'South')) # testing South region

# testing 'North' region
print(region_average(sales_data, 'North')) # no record
```

## Lab 4. Filter by sale amount

```python
# creating a function for filtering sales above threshold
def sales_above(data, threshold): # sales amount that are above threshold
    return [record for record in data if record[2] > threshold] # filtering sales amount
```

```python
# testing sales_data using 500 as threshold
results = sales_above(sales_data, 500)
for record in results: # print line per record
    print(record) #displaying output
```

## BONUS

```python
# creating a function for filtering sales BELOW threshold
def sales_below(data, threshold): # sales amount that are below threshold
    return [record for record in data if record[2] < threshold] # filtering sales amount
```

## Lab 5. Top and bottom sale

```python
# creating a function to find TOP and BOTTOM sales
def top_sale(data): # highest sale amount
    return max(data, key=lambda s: s[2]) # finding highest sale amount tuple (at index 2)
def bottom_sale(data): # lowest sale amount
    return min(data, key=lambda s: s[2]) # finding lowest sale amount tuple (at index 2)
```

```python
# testing to find top and bottom
print(top_sale(sales_data)) # highest sale amount
print(bottom_sale(sales_data)) # lowest sale amount
```

## Lab 6. Apply a discount

```python
# creating a function to discount each sale amount
def apply_discount(data, discount_pct): # discount percentage
    return list(map(lambda s: (s[0], s[1], s[2]*(1 - discount_pct)), data)) # applying and updating each sale amount into new list
```

```python
# testing for 15% discount
discounted = apply_discount(sales_data, 0.15) # discounting 15%

# before discount
print("Before discount:")
print(sales_data[0]) # first record
print(sales_data[1]) # second record
print(sales_data[2]) # third record

# after discount
print("\nAfter 15% discount:")
print(discounted[0]) # first record
print(discounted[1]) # second record
print(discounted[2]) # third record
```

## Lab 7. Unique regions

```python
# creating a function to take unique regions and sort them
def unique_regions(data): # taking unique regions in dataset
    return sorted(set(s[1] for s in data)) # taking regions in first index and deduplicating
```

```python
# testing for unique_regions
print(unique_regions(sales_data))
```

This function helps take data in a larger scale and trim it down so it is easier to analyze. Having duplicates can make it difficult to analzye, and being able to bring up values that are unique can alter the findings in many ways depending on what you are looking for.

## Lab 8. Region summary

```python
# creating a funtion to summarize regions and their sales
def region_summary(data, region): # summary per region
    region_data = [s for s in data if s[1] == region] # filtering by region

    total = region_total(region_data, region) # total sales per region
    count = region_count(region_data, region) # number of sales per region
    average = region_average(region_data, region) # average sale per region

    top = top_sale(region_data) # highest sale amount by region

    # print format
    print(f"=== {region} Region ===")
    print(f"Records: {count}")
    print(f"Total Sales: ${total:,.2f}")
    print(f"Average Sale: ${average:,.2f}")
    print(f"Top Sale: {top[0]} — ${top[2]:,.2f}")
```

```python
# testing each region
region_summary(sales_data, "East") # East region
region_summary(sales_data, "South") # South region
region_summary(sales_data, "West") # West region
```

## Lab 9. Full summary

```python
# creating a loop for every region summary (without using region names)
for r in unique_regions(sales_data): # looping without using region names
    region_summary(sales_data, r) # pulling summary per region
    print() #blank line
```
