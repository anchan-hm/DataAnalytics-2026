## Lab 2. Working with datetime functions
This notebook will be working with dates and times from Python's module

```python
#importing datetime module
import datetime
```

# Today

```python
#creating variable 'today' for current date and time
today = datetime.datetime.now()
print(today) #displaying output in this format to see it better
today #showing raw output
```

```python
#identifying the pieces of information
print(today.strftime("Today is %A, %B %d, %Y")) 
# %A & %B (day of the week & month), %d (day), %Y (4-digit year)
```

# My birthday

```python
#creating variable for my birthday
my_birthday = datetime.date(1998, 4, 20) 
print(my_birthday) #displaying easier output
my_birthday #raw output
```

```python
#formatting birthday as MM/DD/YYYY
print("My birthday is", my_birthday.strftime("%m/%d/%Y"))
```

# Next 90 Days

```python
#creating variable to represent 90 days from today
ninety_d = today + datetime.timedelta(days=90)
print(ninety_d) #easier output
ninety_d #raw output
```

```python
#formatting 90 days from today
print("90 days from today is", ninety_d.strftime("%B %d, %Y"))
```

# Time for Dinner

```python
#varaible for dinner_time
dinner_time = datetime.time(16,20) #example time for 4:20 PM
print(dinner_time) #easier output
dinner_time
```

```python
#formatting dinner_time
print("Let's meet for dinner at", dinner_time.strftime("%I:%M %p"))
```
