## Lab 1. Working with names - parsing strings

```python
test_name = 'John Doe' #creating test name
```

```python
def parse_and_display(name): #defining parse_and_display
    for i in range(len(name)): #grabbing the range in the name
        if name[i] == ' ': #setting up for iterations
            first = name[:i] #first name iteration
            last = name[i+1:] #last name iteration
            print(f"Full name: {name}") #displaying full name
            print(f"First name: {first}") #dispalying first name of the set
            print(f"Last name: {last}") #displaying last name of the set
```

```python
parse_and_display(test_name) #displaying the names and which part they belong to
```

```python
test_list = ['Grace Flores', 'JB Oinonen', 'Ken Penn'] #creating list of names
```

```python
parse_and_display(test_list[0]) #output for Grace Flores [0]
```

```python
parse_and_display(test_list[1]) #ouput for JB Oinonen [1]
```

```python
parse_and_display(test_list[2]) #output for Ken Penn [2]
```

```python
def parse_and_display2(name):
    parts = name.split()

    if len(parts) == 1:
        first = parts[0] #names that are one word
        print(f"Full name: {name}") #showing the full name regardless of number of parts
        print(f"First name: {first}") #showing the first name of each set
        print(f"Middle name: {None}") #if only one name, do not look for any
        print(f"Last name: {None}") #if only one name, do not look for any

    elif len(parts) == 2:
        first, last = parts #first and last names only
        print(f"Full name: {name}") #showing full name
        print(f"First name: {first}") #showing first name of set
        print(f"Middle name: {None}") #showing no middle name of set
        print(f"Last name: {last}") #showing last name of set

    elif len(parts) == 3:
        first, middle, last = parts #first, middle and last names
        print(f"Full name: {name}") #showing full name 
        print(f"First name: {first}") #showing first name of set
        print(f"Middle name: {middle}") #showing middle name of set
        print(f"Last name: {last}") #showing last name of set
    else:
        print(f"No names: {name}") #if names do not fit criteria, the output will be displayed as 'No names'
```

```python
parse_and_display2('Drake') #single name
```

```python
parse_and_display2('Hannah Montana') #first and last name
```

```python
parse_and_display2('Megan Thee Stallion') #first, middle and last names
```
