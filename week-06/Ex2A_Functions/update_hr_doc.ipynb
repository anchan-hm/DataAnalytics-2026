## Lab 3. Employee data
This lab exercise makes updates to information extracted from an employee database.

```python
#creating list with employees
hr_list = [('0123', 'HR', 'Rebecca Yang', '69000'),
           ('0121', 'IT', 'Mark Blick', '67000'),
           ('0068', 'IT', 'Kahna Larsen', '112000'),
           ('0234', 'OPS', 'Jim Smith', '54000')]
```

```python
#updating Mark's name
update = [] #new list for updates
for employee in hr_list:
    if employee[0] == '0121': #Mark's employee number
        new_employee = (employee[0], employee[1], 'Mark Blick-Hawley', employee[3]) #new name
        update.append(new_employee) #adding updates
    else:
        update.append(employee) #adding employee
hr_list = update #saving changes
print(hr_list) #displaying output
```

```python
#updating Jim's job info
update = [] #updates list
for employee in hr_list:
    if employee[0] == '0234': #Jim's employee number
        new_employee = (employee[0], 'CS', employee[2], '60000') #updating department and pay
        update.append(new_employee) # adding updates
    else:
        update.append(employee) #adding employee
hr_list = update #saving changes
print(hr_list) #displaying output
```

```python
#displaying updated contents on multiple lines
for employee in hr_list:
    employee_num = employee[0]
    department = employee[1]
    name = employee[2]
    salary = int(employee[3]) #int to format with commas
    print(f"{employee_num} | {department} | {name} | {salary:,}")
```
