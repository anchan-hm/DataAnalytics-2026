## Lab 2. Parsing strings - parts codes

```python
def get_supplier_code(part_code): #defining get_supplier_code
    for i in range(len(part_code)): #going through the range of index
        if part_code[i] == ':': #going part code interation for colon
            return part_code[:i] #returning anything before the colon
```

```python
def get_product_num(part_code): #defining get_product_num
    colon_pos = part_code.find(':') #finding the colon position
    dash_pos = part_code.find('-') # finding dash position
    return part_code[colon_pos + 1 : dash_pos] #returning what is between the colon and dash
```

```python
def get_size(part_code): #defining get_size
    dash_pos = part_code.find('-') #finding the dash position
    return part_code[dash_pos + 1 :] #returning what is after the dash
```
