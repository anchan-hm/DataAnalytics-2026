```python
# importing library
import numpy as np
```

## Array Basics

```python
# creating a list for fruits
fruits = ["apple","banana","cherry","date"]
fruits
```

```python
# turning list into array
arr_1 = np.array(fruits) # 1D
print(arr_1)
```

```python
# checking data type
arr_1.dtype
```

Unicode string maxed at 6

```python
# creating second array
arr_2 = np.array([[10,11,12],
                  [13,14,15]]) # 2D
print(arr_2)
```

```python
# transposed version of arr_2
print(arr_2.T)
```

```python
# checking to see if arr_2 changed
print(arr_2)
```

arr_2 stayed the same

```python
# creating array with odd and even numbers
arr_3 = np.array([[1,3,5],
                  [2,4,6]]) # 2D
print(arr_3)
```

```python
# slicing arr_3
arr_4 = np.array([arr_3[0,2], arr_3[1,1]]) # grabbing elements (1D)
print(arr_4)
```

```python
# creating array with 1's with 2 values
arr_5 = np.ones((2,2,2)) # 3D
print(arr_5)
```
