## Lab 3. Tax Functions
This notebook defines creating custom functions to calculate taxes for Social Security, Medicare, 
and federal witholding based on a code. This demonstrates using functions, parameters,
return values, and conditional logic.

```python
# creating a function to calculate gross pay for social security tax
def get_soc_sec_tax(gross_pay): #defining social security tax
    return gross_pay*0.062 # 6.2%
```

```python
# creating function to calculate gross pay for medicare tax
def get_medicare_tax(gross_pay): #defining medicare tax
    return gross_pay*0.0145 # 1.45%
```

```python
# creating function to calculate gross pay and withholding code for federal tax
def get_federal_tax(gross_pay, code): #defining federal tax
    if code == 0: #withholding code 0
        rate = 0.23 # 23%
    elif code == 1: #withholding code 1
        rate = 0.21 # 21%
    elif code == 2: #withholding code 2
        rate = 0.195 # 19.5%
    elif code == 3: #withholding code 3
        rate = 0.185 # 18.5%
    else: #withholding code 4+
        rate = 0.18 - ((code - 4) * 0.005) # 18% less 0.5% for each code over 4
    return gross_pay*rate #returning final calculation
```

```python
# testing Person 1
gross = 750 #gross pay
code = 0 #withholding code

#displaying outputs
print("Social Security Tax:", get_soc_sec_tax(gross))
print("Medicare Tax:", round(get_medicare_tax(gross), 2))
print("Federal Tax:", get_federal_tax(gross, code))
```

```python
# testing Person 2
gross2 = 1550 #gross pay
code2 = 2 #withholding code

#displaying outputs
print("Social Security Tax:", get_soc_sec_tax(gross2))
print("Medicare Tax:", round(get_medicare_tax(gross2), 2))
print("Federal Tax:", get_federal_tax(gross2, code2))
```
