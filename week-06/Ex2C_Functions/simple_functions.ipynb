## Lab 1. Working with custom functions
This notebook defines custom functions using parameters and returns.

```python
#creating function for 3 parameters
def favorite_things(name, movie, music): #defining favorite_things
    return f"My name is {name}.\nMy favorite movie is {movie}.\nMy favorite band is {music}." #returning 3 lines and '\n' for line breaking

#displaying output
print(favorite_things("Alexus", "Iron Man", "Falling in Reverse"))
```

```python
#creating function for purpose
def why_im_here(purpose): #defining why_im_here
    return purpose #returning the line

#displaying output
print(why_im_here("I joined Year Up United to build a strong set of skills as a Data Analyst, and be able to have a starting to place pursue Data Science."))
```

```python
#creating function for 2 parameters
def favorite_place(place, why): #defining favorite place and reason
    return f"One of my favorite places is {place} because {why}." #returning line for place and why
```

```python
#displaying output (test 1)
print(favorite_place("Japan", "I loved living there for a period & would want to go back for nostalgia"))
```

```python
#displaying output (test 2)
print(favorite_place("Emerald Isle Beach", "it is one of the nicest beaches in North Carolina"))
```
